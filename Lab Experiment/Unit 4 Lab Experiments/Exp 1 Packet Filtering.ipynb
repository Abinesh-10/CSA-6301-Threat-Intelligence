```python
def handle_outbound_syn(packet, state_table, allow_rules):
    key = (packet["dst"], packet["dport"])

    if key in allow_rules:
        state_table[
            (
                packet["src"],
                packet["sport"],
                packet["dst"],
                packet["dport"],
            )
        ] = "ESTABLISHED"
        return "allow"

    return "deny"


def handle_inbound_packet(packet, state_table):
    key = (
        packet["dst"],
        packet["dport"],
        packet["src"],
        packet["sport"],
    )

    if key in state_table:
        return "allow"

    return "deny"


state_table = {}

allow_rules = {
    ("93.184.216.34", 443)
}

outbound = {
    "src": "10.0.0.20",
    "sport": 51000,
    "dst": "93.184.216.34",
    "dport": 443,
}

print("Outbound:", handle_outbound_syn(outbound, state_table, allow_rules))

response = {
    "src": "93.184.216.34",
    "sport": 443,
    "dst": "10.0.0.20",
    "dport": 51000,
}

print("Inbound Response:", handle_inbound_packet(response, state_table))

unsolicited = {
    "src": "203.0.113.9",
    "sport": 4444,
    "dst": "10.0.0.20",
    "dport": 51000,
}

print("Unsolicited:", handle_inbound_packet(unsolicited, state_table))
print("Exercise 2 completed successfully.")
```
