```python
def rule_matches(packet, rule):
    def field_ok(value, rule_value):
        return rule_value == "any" or value == rule_value

    return (
        field_ok(packet["src"], rule["src"])
        and field_ok(packet["dst"], rule["dst"])
        and field_ok(packet["port"], rule["port"])
        and field_ok(packet["proto"], rule["proto"])
    )


def evaluate_packet(packet, rules):
    for rule in rules:
        if rule_matches(packet, rule):
            return rule["action"]
    return "deny"


misordered_rules = [
    {
        "action": "allow",
        "src": "any",
        "dst": "any",
        "port": "any",
        "proto": "any",
    },
    {
        "action": "deny",
        "src": "10.0.0.5",
        "dst": "any",
        "port": "any",
        "proto": "any",
    },
]

fixed_rules = [
    {
        "action": "deny",
        "src": "10.0.0.5",
        "dst": "any",
        "port": "any",
        "proto": "any",
    },
    {
        "action": "allow",
        "src": "any",
        "dst": "any",
        "port": "any",
        "proto": "any",
    },
]

packet = {
    "src": "10.0.0.5",
    "dst": "8.8.8.8",
    "port": 443,
    "proto": "tcp",
}

print("Misordered Rules:", evaluate_packet(packet, misordered_rules))
print("Corrected Rules:", evaluate_packet(packet, fixed_rules))
print("Default Deny:", evaluate_packet(packet, []))
print("Exercise 1 completed successfully.")
```
