```python
def calculate_risk(likelihood, impact):
    if not (1 <= likelihood <= 5) or not (1 <= impact <= 5):
        raise ValueError(
            "Likelihood and impact must each be between 1 and 5."
        )

    score = likelihood * impact

    if score <= 5:
        level = "Low"
    elif score <= 12:
        level = "Medium"
    elif score <= 19:
        level = "High"
    else:
        level = "Critical"

    return {
        "score": score,
        "level": level
    }


print("Experiment 6: Risk Score Calculator")
print("-" * 55)

# Test Case 1
print("Test Case 1:", calculate_risk(1, 2))

# Test Case 2
print("Test Case 2:", calculate_risk(3, 4))

# Test Case 3
print("Test Case 3:", calculate_risk(4, 4))

# Test Case 4
print("Test Case 4:", calculate_risk(5, 5))

# Test Case 5
print("Test Case 5:", calculate_risk(5, 4))

# Test Case 6 - Invalid input
try:
    calculate_risk(6, 1)
    print("Test Case 6: ERROR")
except ValueError as e:
    print("Test Case 6: ValueError correctly raised")

print("-" * 55)
print("All test cases passed.")
```
